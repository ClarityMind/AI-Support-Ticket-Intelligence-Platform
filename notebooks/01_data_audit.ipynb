```python
body_duplicates = df.duplicated(subset=["body"]).sum()
subject_body_duplicates = df.duplicated(
    subset=["subject", "body"]
).sum()

print(f"Duplicates by body: {body_duplicates}")
print(f"Duplicates by subject + body: {subject_body_duplicates}")
```

```python
body_lengths = df["body"].fillna("").astype(str).str.len()
body_lengths.nlargest(3)
```

```python
body_lengths.nsmallest(5)
```

```python
sum = 0
nan_count = 0
for i in df['body']:
    try:
        sum += len(i)
    except:
        nan_count += 1

print(f"{sum/(len(df['body']) - nan_count):.4f}")
```

```python
arr_median = []
for i in df['body']:
    try:
        arr_median.append(len(i))
    except:
        pass
arr_median.sort()
if len(arr_median) % 2 == 0:
    median = (arr_median[len(arr_median)//2] + arr_median[len(arr_median)//2 - 1]) / 2
else:
    median = arr_median[len(arr_median)//2]

print(median)
```
